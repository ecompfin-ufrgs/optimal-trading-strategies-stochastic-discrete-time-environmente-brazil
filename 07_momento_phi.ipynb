# 07. Momento Phi_chapeu = E[R_p^(1-gamma)]

Só uma função, a `phi_chapeu`, que depois entra na recorrência do A_t. Está em `app/nucleo.py` (requisito F7).

```python
import sys, os
_cwd = os.getcwd()
RAIZ = os.path.dirname(_cwd) if os.path.basename(_cwd) == 'tests' else _cwd
if RAIZ not in sys.path:
    sys.path.insert(0, RAIZ)
import numpy as np
```

## Desenvolvimento

```python
_TOL_FALENCIA = 1e-12
```

```python
def phi_chapeu(alpha, R, rf, gamma):
    """Phi_chapeu = E[R_p^(1-gamma)] da carteira otima (ou E[ln R_p] se gamma=1). (F7)"""
    alpha = np.asarray(alpha, dtype=float)
    excesso = R - rf
    R_p = np.maximum(rf + excesso @ alpha, _TOL_FALENCIA)
    if np.isclose(gamma, 1.0):
        return float(np.mean(np.log(R_p)))
    return float(np.mean(R_p ** (1.0 - gamma)))
```

Teste: se o investidor põe tudo em renda fixa (alpha = 0), o phi tem que dar exatamente R_f elevado a (1 - gamma).

```python
rng = np.random.default_rng(0)
rf_g = 1.003
g = 5.0
R = np.maximum(1+rng.normal(0.01,0.05,200_000),0).reshape(-1,1)
phi_cdi = phi_chapeu(np.array([0.0]), R, rf_g, g)

print('phi (tudo no CDI):', phi_cdi, '| esperado R_f^(1-g):', rf_g**(1-g))
```

```python
assert np.isclose(phi_cdi, rf_g**(1-g))
```
