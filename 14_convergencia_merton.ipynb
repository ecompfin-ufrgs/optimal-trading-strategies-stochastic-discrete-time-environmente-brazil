# 14. Convergência para o resultado de Merton
Validação: quando o intervalo de tempo diminui, o resultado tem que se aproximar da fórmula de Merton. Requisito F14.

```python
import sys, os, tempfile
_cwd = os.getcwd()
RAIZ = os.path.dirname(_cwd) if os.path.basename(_cwd) == 'tests' else _cwd
if RAIZ not in sys.path:
    sys.path.insert(0, RAIZ)
import numpy as np
from app import nucleo
```

**Teste**: quanto menor o intervalo de tempo, menor a distância entre o alpha calculado e o de Merton.

```python
mu_a = 0.10
sig_a = 0.20
rf_a = 0.04
g = 4.0
wm = (mu_a-rf_a)/(g*sig_a**2); rng = np.random.default_rng(0); erros = {}
for nome, dt in [('anual',1.0),('mensal',1/12),('semanal',1/52)]:
    r = rng.normal(mu_a*dt, sig_a*np.sqrt(dt), 500_000)
    a = nucleo.resolver_alpha_otimo(np.maximum(1+r,0).reshape(-1,1), (1+rf_a)**dt, g)
    erros[nome] = abs(a[0]-wm); print(nome, 'alpha*=', round(a[0],4), 'erro=', round(erros[nome],4))
assert erros['mensal'] < erros['anual'] and erros['semanal'] < erros['mensal']
```
