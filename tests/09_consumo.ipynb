**Teste** — c_t = θ_t·W_t e c_t + S_t = W_t.

```python
rng = np.random.default_rng(0); rf_g, g, beta, T = 1.003, 4.0, 0.96, 12
R = np.maximum(1+rng.normal(0.01,0.05,100_000),0).reshape(-1,1)
alpha = nucleo.resolver_alpha_otimo(R, rf_g, g)
theta = nucleo.fracoes_consumo(nucleo.recorrencia_A(nucleo.phi_chapeu(alpha,R,rf_g,g),beta,g,T), g)
Rp = np.maximum(1+rng.normal(0.01,0.05,(2000,T,1)),0)
out = nucleo.propagar_riqueza(1.0, theta, alpha, Rp, rf_g)
print('c_0 medio:', out['c'][:,0].mean())
assert all(np.allclose(out['c'][:,t], theta[t]*out['W'][:,t]) for t in range(T+1))
assert np.allclose(out['c']+out['S'], out['W'])
print('F9 consumo: PASSOU')
```
