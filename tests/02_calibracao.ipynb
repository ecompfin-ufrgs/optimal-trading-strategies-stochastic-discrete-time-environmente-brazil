**Teste**: a partir dos retornos, sai a média, a covariância e o R_f.

```python
rng = np.random.default_rng(0)
ret = pd.DataFrame({'data': pd.date_range('2000-01', periods=120, freq='MS').strftime('%Y-%m'), 'ibov': rng.normal(0.012, 0.05, 120)})
mv = RendaVariavel(ret)
rf = RendaFixa(0.10)
mu = mv.media()
sig = mv.covariancia()
r = rf.retorno_livre_risco()

print('mu_hat:', mu, '| Sigma_hat:', sig, '| R_f:', r)
```

```python
assert mu.shape == (1,) and sig.shape == (1, 1) and r > 0
```
