```python
_TOL_FALENCIA = 1e-12

def phi_chapeu(alpha, R, rf, gamma):
    """Φ̂ = E[R_p^(1−γ)] do portfólio ótimo (ou E[ln R_p] se γ=1). (F7)"""
    alpha = np.asarray(alpha, dtype=float)
    excesso = R - rf
    R_p = np.maximum(rf + excesso @ alpha, _TOL_FALENCIA)
    if np.isclose(gamma, 1.0):
        return float(np.mean(np.log(R_p)))
    return float(np.mean(R_p ** (1.0 - gamma)))
```

**Teste** — all-cash dá Φ̂ = R_f^(1−γ).

```python
rng = np.random.default_rng(0); rf_g, g = 1.003, 5.0
R = np.maximum(1+rng.normal(0.01,0.05,200_000),0).reshape(-1,1)
phi_cash = phi_chapeu(np.array([0.0]), R, rf_g, g)
print('phi (all-cash):', phi_cash, '| esperado R_f^(1-g):', rf_g**(1-g))
assert np.isclose(phi_cash, rf_g**(1-g))
print('F7 phi_chapeu: PASSOU')
```
